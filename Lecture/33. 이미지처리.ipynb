### Numpy를 Image로 변환

```python
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
fruits = np.load("../Data/fruits.npy")
fruits.shape
```

#### Image로 저장하기

```python
from PIL import Image
```

```python
for i in range(300):
    image = Image.fromarray(fruits[i]).convert("L") # L:흑백톤
    image.save("../Data/fruits/fruit_%03d.jpg"%i, "JPEG")
```

----
#### Image를 읽어서 Numpy로 변환하기

```python
import glob
```

```python
numberOfData = 300 # 데이터 갯수
imgSize = 100 # 이미지 사이즈
```

```python
# 배열 초기화
fruitsData = np.zeros(
                    numberOfData * imgSize * imgSize,
                    dtype=np.int32
).reshape(numberOfData, imgSize, imgSize)

fruitsData.shape
```

```python
fruitsData
```

```python
i = 0
for file in sorted(glob.glob("../Data/fruits/*.jpg")):
    img = np.array(Image.open(file), dtype=np.int32)
    fruitsData[i,:,:] = img
    i+=1
```
