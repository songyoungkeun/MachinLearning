#### 사과 이미지 보기 : 이미지로 확인하기

```python
import matplotlib.pyplot as plt
```

```python
plt.imshow(fruits[0], cmap='gray')
plt.show()
```

```python
# 사과, 파인애플, 바나나 이미지를 나란히 보기 

fig, axs = plt.subplots(1, 3, figsize=(20, 5))
axs[0].imshow(fruits[0], cmap='gray_r')
axs[1].imshow(fruits[100], cmap='gray_r')
axs[2].imshow(fruits[200], cmap='gray_r')

plt.show()
```

#### 샘플 차원 변경하기 
- 현재 3차원의 배열을 분석하기 위해 2차원 배열로 변경이 필요
- 배열중 page단의 변수 apple, pineapple, banana로 분류하여 변경

```python
apple = fruits[0:100].reshape(-1,100*100) # (행,렬)에서 열의 기준만 설정하고 행은 알아서 처리
pineapple = fruits[100:200].reshape(-1,100*100)
banana = fruits[200:300].reshape(-1,100*100)

banana.shape
```

#### Pixel값 분석하기

```python
# apple의 sample별 행별 평균 구하기 
apple.mean(axis=1)
```

```python
# 과일별 빈도수를 히스토그램으로 표현하기
plt.figure(figsize=(15, 5))
plt.hist(
    np.mean(apple, axis=1),
    alpha=0.5
)
plt.hist(
    np.mean(banana, axis=1),
    alpha=0.5
)
plt.hist(
    np.mean(pineapple, axis=1),
    alpha=0.5
)

plt.legend(['apple', 'banana', 'pineapple'])

plt.show()
```

> x축은 픽셀의 평균값, y축은 빈도수   
바나나와 다른 과일은 분리가 쉬우나 사과와 파인애플은 분리가 어렵다.   
바나나의 넓이는 적고, 사과와 파인애플의 넓이느 비슷하다.    

#### Feature별(픽셀별) 평균의 막대 그래프 그리기

```python
fig, axs = plt.subplots(1, 3, figsize=(20, 5))
axs[0].bar(range(10000), np.mean(apple, axis=0))
axs[1].bar(range(10000), np.mean(pineapple, axis=0))
axs[2].bar(range(10000), np.mean(banana, axis=0))

plt.show()
```

> x축은 픽셀 갯수 10000개, y축은 각 픽셀의 평균값    
바나나인 경우에는 중앙값이 높다.    
사과는 이미지의 아랫쪽 부분이 높다.    
파인애플은 전반적으로 고르게 분포하고 있다.  

#### 평균 이미지로 그림 표현하기 
: 이미지로 그리기 위해 다시 2차원 배열로 변경

```python
apple_mean = np.mean(apple, axis=0).reshape(100, 100)
pineapple_mean = np.mean(pineapple, axis=0).reshape(100, 100)
banana_mean = np.mean(banana, axis=0).reshape(100, 100)

fig, axs = plt.subplots(1, 3, figsize=(20, 5))
axs[0].imshow(apple_mean, cmap='gray')
axs[1].imshow(pineapple_mean, cmap='gray')
axs[2].imshow(banana_mean, cmap='gray')

plt.show()
```

> 각 과일의 픽셀별 평균을 이미지로 그린 그림이다. 

#### 평균과 가까운 사진 고르기
- 평균 이미지와 각 샘플들의 픽셀값과 비교하여 가까운 거리의 이미지는 해당 과일일 것이다. 

```python
abs_diff = np.abs(fruits - apple_mean)
abs_mean = np.mean(abs_diff, axis=(1,2)) # (0,1,2) 행과 열의 평균값을 구한다. 
# abs_diff가 3차원 이기 때문에 page차원을 제외하고 행과 열만의 평균을 구한다.
abs_mean.shape
```

```python
abs_mean
```

#### 300개의 데이터중 차이가 적은 순으로 정렬하면서 해당 이미지를 불러오자

```python
# data의 적은순으로 정렬하여 index번호 찾기
apple_index = np.argsort(abs_mean)[:100]
apple_index
```

```python
# 그림 그려보기
fig, axs = plt.subplots(10, 10, figsize=(10, 10))

for i in range(10):
    for j in range(10):
        axs[i,j].imshow(fruits[apple_index[i*10+j]], cmap='gray')
        axs[i,j].axis('off')

plt.show()
```

> 이와 같은 방법으로 비지도 학습인 군집분류가 알고리즘으로 구성되어 있다.  

#### 파인애플로 확인해 보기

```python
abs_diff = np.abs(fruits - pineapple_mean)
abs_mean = np.mean(abs_diff, axis=(1,2)) # (0,1,2) 행과 열의 평균값을 구한다. 
# abs_diff가 3차원 이기 때문에 page차원을 제외하고 행과 열만의 평균을 구한다.
abs_mean.shape
```

```python
# data의 적은순으로 정렬하여 index번호 찾기
pineapple_index = np.argsort(abs_mean)[:100]
pineapple_index
```

```python
# 그림 그려보기
fig, axs = plt.subplots(10, 10, figsize=(10, 10))

for i in range(10):
    for j in range(10):
        axs[i,j].imshow(fruits[pineapple_index[i*10+j]], cmap='gray')
        axs[i,j].axis('off')

plt.show()
```
